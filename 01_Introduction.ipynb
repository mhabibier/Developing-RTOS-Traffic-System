## Supervised Learning

Supervised learning adalah metode machine learning yang menggunakan
data berlabel. Setiap data masukan memiliki jawaban atau target yang
sudah diketahui.

Supervised learning umumnya dibagi menjadi:

1. Classification, untuk memprediksi kategori.
2. Regression, untuk memprediksi nilai numerik kontinu.

Contoh classification adalah memprediksi apakah slot parkir kosong
atau terisi. Contoh regression adalah memprediksi jumlah kendaraan
yang datang setiap jam.

```python
from sklearn.datasets import load_iris

iris_dataset = load_iris()

print("Keys:", iris_dataset.keys())
print("Target names:", iris_dataset.target_names)
print("Feature names:", iris_dataset.feature_names)
```

### Penjelasan Hasil

Kode tersebut memuat dataset Iris yang tersedia di scikit-learn.
Dataset ini berisi pengukuran bunga Iris berdasarkan panjang dan
lebar sepal serta petal.

`target_names` menunjukkan tiga kelas bunga yang akan diprediksi,
sedangkan `feature_names` menunjukkan variabel yang digunakan sebagai
masukan model.

## Membagi Data Training dan Testing

Data training digunakan untuk melatih model, sedangkan data testing
digunakan untuk menguji kemampuan model terhadap data yang sebelumnya
belum dilihat.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset.data,
    iris_dataset.target,
    test_size=0.25,
    random_state=42
)

print("Ukuran X_train:", X_train.shape)
print("Ukuran X_test:", X_test.shape)
```

Dataset dibagi menjadi 75% data training dan 25% data testing.
`random_state=42` digunakan agar pembagian data menghasilkan hasil
yang sama ketika notebook dijalankan kembali.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for jumlah_tetangga in [1, 3, 5, 7]:
    model = KNeighborsClassifier(n_neighbors=jumlah_tetangga)
    model.fit(X_train, y_train)

    prediksi = model.predict(X_test)
    akurasi = accuracy_score(y_test, prediksi)

    print(f"n_neighbors={jumlah_tetangga}, akurasi={akurasi:.3f}")
```
